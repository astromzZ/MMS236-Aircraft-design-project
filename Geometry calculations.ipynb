```python
import numpy as np
from scipy.integrate import quad

#Wing Trapazoidal (non cranked)

# c_r = 17.67 # root chord
# c_t = 3.98 #tip chord
# Taper_r = c_t/c_r  #taper ratio
# print('A380 Taper Ratio =',Taper_r)
# b = 79.75 #span

L = 70.40000
A = 10
S = 593.1557831729295
Tank_Volume = 616.4
W_fuel = 39225
M_cruise = 0.85

print('Aspect ratio = ', A )
print('Area ratio = ', S )

# c_t = 3.98 #tip chord

# ct = taper * cr
Taper = 0.2 #taper ratio

b = np.sqrt(A * S) #aspect ratio
print('span=',b)

c_r = ((S*2)/(b*(1+Taper)))
print('chord root =',c_r)
c_t = Taper * c_r
print('chord tip = ', c_t)

Taper = c_t/c_r  #taper ratio
print('Taper Ratio =',Taper)


c_mac = 2/3* c_r * ((1+Taper+Taper**2)/(1+Taper))  #Mean aerodynamic chord
print('Mean aerodynamic chord =',c_mac)

Y_mac = b/6 * (1+2*Taper)/(1+Taper)
print('Y location of mac=',Y_mac)

print(b/2 - Y_mac)


# sweep_rad = np.arctan((c_r-c_t)/(b))
# sweep_rad = np.arctan((46.97-19.24)/((b/2)-7.14/2))
# sweep = sweep_rad*180/np.pi
sweep = 25
print('sweep =',sweep)

wing_tail_dis = 58.443 - 17.400
print(wing_tail_dis)
print(wing_tail_dis/L)

Fuselage_cg_x_loc = 0.45 * L
print('Cg x distace from fuselage nose =', Fuselage_cg_x_loc)

#due to OpenVsp, need to 

Wing_PoR_to_LeMac = Y_mac*np.tan(sweep*np.pi/180)
print('Wing_PoR_to_LeMac =',Wing_PoR_to_LeMac)

wing_loc_rel_cg = Fuselage_cg_x_loc - (Wing_PoR_to_LeMac + 0.3*c_mac)

print('wing_loc_mac_rel_cg = ', wing_loc_rel_cg)
print('nose to leading edge mean chord =', wing_loc_rel_cg - 0.3*c_mac)

##airfoil T/c

#standerd = 0.10000
T_c_tip = 0.1295 #historically using M = 0.85 
print('t/c tip =', T_c_tip)

T_c_root = (1+.6)*T_c_tip
print('t/c root =', T_c_root)

T_c_mid = (T_c_root+T_c_tip)/2
print('t/c mid =',T_c_mid)
c = 2

print(c*T_c_tip)
print
```
